### D9. Loop over a dict

Using `record` from D7, write three loops:

### D10. Build a dict from two lists with `zip()`

### C1. Load records safely from a file

Write `load_records(path)` with a one-line docstring. It reads `label,value,limit` lines from a file
and returns a **list of dicts**. If the file does not exist, it prints a clear message and returns an
empty list instead of crashing.

Then load `records_sample.txt` and print each record's label and status, as in D16.

**Test it:**

| Call | Expected output |
|---|---|
| `load_records("records_sample.txt")` | the same three lines as D16 |
| `load_records("nope.txt")` | a clear message, then `[]` |

````python
# C1
def load_records(path):
    """Read label,value,limit lines from a file and return a list of dicts."""
    records = []
    try:
        with open(path) as f:
            for line in f:
                line = line.strip()
                if line == "":
                    continue
                label, value, limit = line.split(",")
                records.append({"label": label,
                                "value": float(value),
                                "limit": float(limit)})
    except FileNotFoundError:
        print(f"Could not find {path}")
        return []
    return records


records = load_records("records_sample.txt")
for record in records:
    if record["value"] > record["limit"]:
        status = "OVER LIMIT"
    else:
        status = "OK"
    print(f"{record['label']}: {status}")

print(load_records("nope.txt"))
````

Using your records from C1, write one line per record to a new file called `results.txt`,

**Expected output:**
```

```

## What to build

**Threshold — pass standard.** Write your records as a **list of dicts directly in
the code** — at least 3. Bring back `check_record` and `print_report` from Week 3.
Loop over your records and print a report for each.

**Typical.** Instead of writing the list in the code, **load it from a file** —
write `load_records(path)`. Make your own data file, one record per line,
`label,value,limit`.

**Excellent.** Also **write the results out to a new file** — one line per record,
`label,status` — as well as printing them. Keep a running count of how many were
`OVER LIMIT` and print that once, after the loop ends. Wrap the call to
`load_records` in a `try`/`except` so a missing file prints a clear message and
starts with an empty list, instead of crashing (Topic 4).

## Rules

- **Every function does one job**, with a one-line docstring.
- Do not write the percentage/status logic more than once.
- Loading and processing are separate steps — `load_records` should not decide
  any status itself, only build the list.

## When it works

1. Run it and check every record's numbers by hand.
2. If you are loading from a file, try it with a file that does not exist. If
   you're on Threshold/Typical, write the crash down — that's next. If you're on
   Excellent, your `try`/`except` should catch it instead.

````python
"""
RECORD CHECK  -  my version
===========================

Name  :  Victor Dzingai
Lane  :  Cyber Security
Date  :  15/10/2026

Run it:   python template.py

Each line of failed_logins.txt is: source IP, failed logins, allowed failed logins.
For example:   10.0.0.5,12,10
"""

# =================================================================== FUNCTIONS
# Each function does ONE job and has a one-line docstring.


def check_record(value, limit):
    """Return the difference, percent and status for one record."""
    # This is the only place the percentage and status are worked out.
    difference = value - limit
    percent = (value / limit) * 100
    if value > limit:
        status = "OVER LIMIT"
    elif percent >= 90:
        status = "WARNING"
    else:
        status = "OK"
    return difference, percent, status


def print_report(label, value, limit, difference, percent, status):
    """Print the report for one record."""
    # :>5.1f   right-aligned, 1 decimal place
    # :>+6.1f  the same, but always shows the + or - sign
    print(f"{label:<12} failed: {value:>5.1f}  limit: {limit:>5.1f}  "
          f"diff: {difference:>+6.1f}  {percent:>6.1f}%  {status}")


def print_summary(over_count):
    """Print how many records were over the limit."""
    print(f"\nRecords over limit: {over_count}")


# ====================================================================== STORE
# Each record is a dictionary: {"label": ..., "value": ..., "limit": ...}
# All the records together go in a list.


def load_records(path):
    """Read label,value,limit lines from a file and return a list of dicts."""
    # This only builds the list. It does not decide any status.
    records = []
    # "with" closes the file for us when we are finished with it
    with open(path) as f:
        for line in f:
            line = line.strip()          # remove the new line at the end
            if line == "":
                continue                 # skip empty lines
            # "10.0.0.5,12,10" is split at the commas into three pieces
            label, value, limit = line.split(",")
            # the numbers are text in a file, so convert them with float()
            records.append({"label": label,
                            "value": float(value),
                            "limit": float(limit)})
    return records


def write_results(results, path):
    """Write one 'label,status' line per result to a file."""
    # "w" means write. It makes a new file, or replaces the old one.
    with open(path, "w") as f:
        for label, status in results:
            f.write(f"{label},{status}\n")    # \n starts a new line


# =============================================================== MAIN PROGRAM
# If the file is missing, load_records crashes with FileNotFoundError.
# try / except catches that, prints a message, and carries on with no records.
try:
    records = load_records("failed_logins.txt")
except FileNotFoundError:
    print("Could not find failed_logins.txt - starting with no records.")
    records = []

results = []       # will hold (label, status) pairs to write to the file
over_count = 0     # starts at 0 before the loop

for record in records:
    # Get the answers back from check_record, then print the report.
    difference, percent, status = check_record(record["value"], record["limit"])
    print_report(record["label"], record["value"], record["limit"],
                 difference, percent, status)

    results.append((record["label"], status))
    if status == "OVER LIMIT":
        over_count += 1

# These two lines are outside the loop, so they run once at the end.
print_summary(over_count)
write_results(results, "login_results.txt")
````
