### D6. A default value

1. Define `status_of(percent, warning_at=90)` that returns:
   - `"OVER LIMIT"` if `percent` is 100 or more
   - `"WARNING"` if `percent` is `warning_at` or more
   - `"OK"` otherwise
2. Print `status_of(95)`, then `status_of(85)`, then `status_of(85, 80)`.

**Expected output:** `WARNING`, `OK`, `WARNING`

````python
# D6
def status_of(percent, warning_at=90):
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"

print(status_of(95))
print(status_of(85))
print(status_of(85, 80))
````

### D8. `*args` and `**kwargs`

a) Define `total(*numbers)` that returns the sum of all the numbers passed in. Print `total(1, 2, 3)`,

**Expected output:**

````python
# D8
def total(*numbers):
    return sum(numbers)

print(total(1, 2, 3))
print(total(10, 20, 30, 40))
print(total())


def configure(**settings):
    for key, value in settings.items():
        print(f"{key} = {value}")

configure(host="10.0.0.1", port=8080)
````

### C2. How many are over?

Write `count_over(*values, limit=100)` with a one-line docstring. It returns how many of the values
are over the limit.

| Call | Expected output |
|---|---|
| `count_over(87, 120, 45, 101)` | `2` |
| `count_over(87, 120, 45, 101, limit=50)` | `3` |
| `count_over()` | `0` |

````python
# C2
def count_over(*values, limit=100):
    """Return how many of the values are over the limit."""
    count = 0
    for value in values:
        if value > limit:
            count += 1
    return count

print(count_over(87, 120, 45, 101))
print(count_over(87, 120, 45, 101, limit=50))
print(count_over())
````

## What to build

**Threshold — pass standard.** Write `status_of(percent)`. Ask for the three values,
work out the difference and percentage inline, call `status_of` for the status, and
print a bordered report including it.

```
==================================
  RECORD CHECK  -  srv-01
==================================
  Used        : 87
  Total       : 120
  Status      : OK
==================================
```

**Typical.** Also write `check(value, limit)`, returning the difference and

```
==================================
  RECORD CHECK  -  srv-01
==================================

  Total       :     120.00

```

**Excellent.** Also write `print_report(...)` — every `print()` for the report
happens inside it, nowhere else. Wrap the input/process/output steps in a loop so you
can check as many records as you like in one run — type `quit` to stop. Keep count of
how many came back `OVER LIMIT` and print that once, after the loop ends.

## Rules

- **Every function does one job.** If a function both calculates and prints, split it.
- Do not calculate the same thing in two different places.
- **Give every function you write a one-line docstring** saying what it does.
- **No lists yet** — that is Week 4.

## When it works

1. Run it three times with different inputs. Does it still look right?
2. Run it with a total of `0`. Write the error down.

````python
"""
RECORD CHECK  -  my version
===========================

Name  :  Victor Dzingai
Lane  :  Cyber Security
Date  :  08/10/2026

Run it:   python template.py

Each record is: source IP, failed logins, allowed failed logins (the limit).
Type quit as the source IP to stop.
"""

# =================================================================== FUNCTIONS
# A function is a named block of code that we can use again and again.
# Each one below does ONE job, and has a one-line docstring saying what it does.


def status_of(percent):
    """Return OVER LIMIT, WARNING or OK for a percentage."""
    # return sends the answer back to the line that called the function
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= 90:
        return "WARNING"
    else:
        return "OK"


def check(value, limit):
    """Return the difference and the percentage of value against limit."""
    # This only calculates. It does not print anything.
    difference = value - limit
    percent = (value / limit) * 100
    # A function can return more than one value. They come back in this order.
    return difference, percent


def print_report(label, value, limit, difference, percent, status):
    """Print the bordered report for one record."""
    # All of the printing for the report happens here and nowhere else.
    # The function can only use what we pass in, so everything is a parameter.
    print()
    print("=" * 34)
    print(f"  RECORD CHECK  -  {label}")
    print("=" * 34)
    print(f"  Failed      : {value:>10.2f}")
    print(f"  Allowed     : {limit:>10.2f}")
    print(f"  Difference  : {difference:>+10.2f}")
    print(f"  Percent     : {percent:>10.2f} %")
    print(f"  Status      : {status:>10}")
    print("=" * 34)


# ================================================================ MAIN PROGRAM
# This counter starts at 0 before the loop so it is not reset on every pass.
over_limit_count = 0

while True:

    # INPUT
    source_ip = input("Source IP (or quit to stop): ")
    if source_ip == "quit":
        break

    # input() gives back text, so the numbers are converted with float()
    failed_logins = float(input("Failed logins: "))
    allowed_failed = float(input("Allowed failed logins: "))

    # PROCESS
    # The two values that check() returns are put into two variables.
    # They must be in the same order as the return line in check().
    difference, percent = check(failed_logins, allowed_failed)
    status = status_of(percent)

    # OUTPUT
    print_report(source_ip, failed_logins, allowed_failed, difference, percent, status)

    if status == "OVER LIMIT":
        over_limit_count += 1

# Not indented, so this only runs once after the loop has finished.
print()
print(f"Records over limit: {over_limit_count}")
````
